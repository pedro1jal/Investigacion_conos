```python
#\(f(x,y)=20+(x^{2}-10\cos (2\pi x))+(y^{2}-10\cos (2\pi y))\)
import numpy as np

# def f(x):
#     x = np.array(x)
#     return 20 + (x[0]**2 - 10*np.cos(2*np.pi*x[0]))+ (x[1]**2 - 10*np.cos(2*np.pi*x[1]))

def f(x):
    """
    Función de Ackley en 2D.
    x: array-like o vector de NumPy con al menos 2 elementos (x[0], x[1])
    """
    x0 = x[0]
    x1 = x[1]
    
    term1 = -20.0 * np.exp(-0.2 * np.sqrt(0.5 * (x0**2 + x1**2)))
    term2 = -np.exp(0.5 * (np.cos(2 * np.pi * x0) + np.cos(2 * np.pi * x1)))
    
    return term1 + term2 + 20 + np.e

def grad(x):
    """
    Gradiente analítico de la función de Ackley en 2D.
    Devuelve un vector de NumPy [df/dx0, df/dx1].
    """
    x0 = x[0]
    x1 = x[1]
    
    # Manejo del caso especial en el origen para evitar división por cero en la raíz
    if x0 == 0 and x1 == 0:
        return np.array([0.0, 0.0])
    
    # Pre-cálculo de componentes comunes para optimizar velocidad
    r2 = x0**2 + x1**2
    sqrt_r2 = np.sqrt(0.5 * r2)
    exp1 = np.exp(-0.2 * sqrt_r2)
    exp2 = np.exp(0.5 * (np.cos(2 * np.pi * x0) + np.cos(2 * np.pi * x1)))
    
    # Derivadas parciales con respecto a x0 y x1
    df_dx0 = (2.0 * x0 * exp1) / sqrt_r2 + 0.5 * np.pi * np.sin(2 * np.pi * x0) * exp2
    df_dx1 = (2.0 * x1 * exp1) / sqrt_r2 + 0.5 * np.pi * np.sin(2 * np.pi * x1) * exp2
    
    return np.array([df_dx0, df_dx1])
lim = 32.768
def proyectar_en_espacio(punto):
    x, y = punto[0], punto[1]
    # Restringe x e y para que no salgan de las rectas frontera
    x_proyectado = np.clip(x, -lim,lim)
    y_proyectado = np.clip(y, -lim,lim)
    return np.array([x_proyectado, y_proyectado])

def descenso_gradiente(f,grad, x0, alpha=0.0002, tol=1e-5, max_iter=2000):
    x = np.array(x0)
    k = 0
    for i in range(max_iter):
        k += 1
        grad_f = grad(x)
        # print(grad)
        x_new = x - alpha * grad_f
        x_new = proyectar_en_espacio(x_new)
        if np.linalg.norm(x_new - x) < tol:
            x = x_new
            break
        # if np.linalg.norm(grad_f) < tol:
        #     x = x_new
        #     break
        
    return x ,k 

x0 = np.array([-90, -80])
epsy = 1e-5
mejor_x = x0
mejor_y = f(x0)
y_star1 = f(mejor_x)
for it in range(10):
    print(f"Iteración n°: {it+1}")
    x_star,k = descenso_gradiente(f,grad, x0, alpha=0.0002, tol=1e-4, max_iter=2000)
    print(f"Iteraciones: {k}")
    y_star2 = f(x_star)
   
    # np.random.seed(it)
    if np.abs(y_star2 - y_star1) < epsy:
        print(f"Convergencia alcanzada en la iteración {it+1}")
        break
    x_sample = np.random.uniform(-lim, lim, size=(5000, 2))
    lista = [(x , f(x)) for x in x_sample]
    min_lista = min(lista , key=lambda item: item[1])
    # print(min_lista, min_lista[1])
    
    if f(x_star) <= min_lista[1]:
        print(f"Mínimo global: {np.round(x_star,3)} - f(x): {np.round(f(x_star),3)}")
        break
    else:
        # print(min_lista[1])
        x0 = min_lista[0]
        print(f"minimo local: {np.round(x0,3)} - f(x): {np.round(f(x0),2)}")
```
